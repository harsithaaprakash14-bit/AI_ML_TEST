```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Generate 1000 records
n_records = 1000

data = {
    'customer_id': range(1001, 1001 + n_records),
    'age': np.random.randint(18, 70, size=n_records),
    'annual_income': np.random.randint(20000, 150000, size=n_records),
    'credit_score': np.random.randint(300, 850, size=n_records),
    'loan_amount': np.random.randint(5000, 80000, size=n_records),
    'loan_term_months': np.random.choice([12, 24, 36, 48, 60], size=n_records)
}

# Calculate a basic logic-based loan status (Approved / Rejected) to make it look realistic,
# where higher credit score and income increase approval chances.
df = pd.DataFrame(data)

score_factor = (df['credit_score'] / 850) * 0.5 + (df['annual_income'] / 150000) * 0.5
status_threshold = df['loan_amount'] / 100000

df['loan_status'] = np.where(score_factor > status_threshold * 0.4, 'Approved', 'Rejected')

# Display the shape and first few rows
print(f"Dataset Shape: {df.shape}")
display(df.head())

# Show the distribution of Approved vs Rejected
print("\nLoan Status Distribution:")
print(df['loan_status'].value_counts())
```

```python
import pandas as pd
import numpy as np

# Create a specific dataset with exactly 6 records containing both Approved and Rejected cases
data_6 = {
    'customer_id': [1001, 1002, 1003, 1004, 1005, 1006],
    'age': [25, 45, 30, 58, 22, 50],
    'annual_income': [35000, 120000, 55000, 95000, 28000, 110000],
    'credit_score': [580, 780, 620, 810, 450, 710],
    'loan_amount': [15000, 30000, 40000, 10000, 20000, 85000],
    'loan_term_months': [36, 12, 48, 24, 36, 60],
    'loan_status': ['Approved', 'Approved', 'Rejected', 'Approved', 'Rejected', 'Rejected']
}

df = pd.DataFrame(data_6)

print("New Dataset (6 records):")
display(df)
print("\nLoan Status Distribution:")
print(df['loan_status'].value_counts())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# Features and target encoding
X = df[['age', 'annual_income', 'credit_score', 'loan_amount', 'loan_term_months']]
y = df['loan_status'].map({'Approved': 1, 'Rejected': 0})

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Since the dataset is extremely small (6 records), we train on the entire set to get the best decision boundary
model = LogisticRegression(random_state=42)
model.fit(X_scaled, y)

y_pred = model.predict(X_scaled)

print("Training Evaluation on 6-record Dataset:")
print(classification_report(y, y_pred, target_names=['Rejected', 'Approved']))
print("Accuracy Score:", accuracy_score(y, y_pred))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

# Compute confusion matrix
cm = confusion_matrix(y, y_pred)
tn, fp, fn, tp = cm.ravel() if cm.size == 4 else (cm[0][0], 0, 0, 0) # Fallback handling for edge cases

# Prepare classification outcomes for the pie chart
labels = ['True Negatives (Correct Rejections)', 'False Positives', 'False Negatives', 'True Positives (Correct Approvals)']
sizes = [tn, fp, fn, tp]
colors = ['#ff9999','#ffcccc','#99ff99','#66b3ff']

# Keep only classes that exist
filtered_labels = [label for label, size in zip(labels, sizes) if size > 0]
filtered_sizes = [size for size in sizes if size > 0]
filtered_colors = [color for color, size in zip(colors, sizes) if size > 0]

plt.figure(figsize=(6, 6))
plt.pie(filtered_sizes, labels=filtered_labels, colors=filtered_colors,
        autopct='%1.1f%%', shadow=True, startangle=140)
plt.title('Outcome Distribution on 6-record Dataset')
plt.show()
```
